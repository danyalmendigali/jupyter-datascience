## Pearson correlation and cousine similarity 

```python
def CorrAndCous(a, b):
    dt = np.dot(a, b)
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    cos = (dt / (norm_a * norm_b))


    x_avg = a - np.mean(a)
    y_avg = b - np.mean(b)

    g1 = np.dot(x_avg, y_avg)
    g2 = np.linalg.norm(x_avg) * np.linalg.norm(y_avg)

    pirson = g1 / g2

    return cos, pirson


v1 = np.array([1, 2, 3])
v2 = np.array([1, 2, 3])

f, f2 = CorrAndCous(v1, v2)

print(f, f2)
print(v1)
print(v2)

print(np.corrcoef(v2))
    
```

# k-means 

```python
np.random.seed(42)
data = np.random.randn(150, 2)

plt.scatter(data[:,0], data[:,1])
plt.show()
```

```python
np.random.seed(42)
data = np.random.randn(150, 2)

k = 4
iterations = 4
ridx = np.random.choice(range(len(data)), k, replace=False)
centroids = data[ridx, :]


plt.figure(figsize=(12, 3))

for it in range(iterations + 1):
    dists = np.zeros((data.shape[0], k))
    for ci in range(k):
        dists[:, ci] = np.sum((data - centroids[ci,:])**2, axis=1)
    
    groupidx = np.argmin(dists, axis=1)
    
    plt.subplot(1, iterations + 1, it + 1)
    plt.scatter(data[:,0], data[:,1], c=groupidx, cmap='viridis', s=30)
    plt.scatter(centroids[:,0], centroids[:,1], color='red', marker='X', s=100)
    plt.title(f"Iteration {it}")
    plt.xticks([])
    plt.yticks([])
    
    if it < iterations:
        for ki in range(k):
            centroids[ki, :] = np.mean(data[groupidx == ki], axis=0)

plt.tight_layout()
plt.show()
```
