## distributive property(aT (b + c) = aTb + aTc)

```python
a = np.array([0, 1, 2])
b = np.array([3, 5, 8])
c = np.array([13, 21, 34])
res1 = np.dot(a, b + c)
res2 = np.dot(a, b) + np.dot(a, c)
print(res1, res2)
```

```python
v = np.array([1, 2])
w = np.array([4, -6])
vPlusW = v + w

plt.figure(figsize=(6,6))
plt.axis('equal')
plt.grid(True)
plt.xlim(-1, 6)
plt.ylim(-7, 6)

plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(v[0], v[1], w[0], w[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, vPlusW[0], vPlusW[1], angles='xy', scale_units='xy', scale=1)
```

```python
def n(a):
    return np.sqrt(np.sum(a ** 2))

a = np.array([1, 2, 3])
norm = np.linalg.norm(a)
print(n(a))
print(norm)
```

```python
def ei(v):
    return (1 / np.linalg.norm(v) * v)

g = np.array([1, 2, 3])
g_norm1 = np.linalg.norm(ei(g))
print(ei(g))
print(g_norm1)
```
