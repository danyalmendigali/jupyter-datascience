```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import matplotlib_inline.backend_inline
import plotly.graph_objects as go
import plotly.io as pio
matplotlib_inline.backend_inline.set_matplotlib_formats('svg') 
plt.rcParams.update({'font.size':14})  
```

```python
l1 = 1 
l2 = 2 
l3 = -3

v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
v3 = np.array([7, 8, 9])

l1 * v1 + l2 * v2 + l3 * v3
```

```python
scalars = [l1, l2, l3]
vectors = [v1, v2, v3]

vec = np.zeros(len(v2))

for s, v in zip(scalars, vectors):
    vec += s * v

vec
```

```python
scalars = [l1, l2, l3, 5]
vectors = [v1, v2, v3]

vec = np.zeros(len(v2))

for i in range(len(scalars)):
    vec += scalars[i] * vectors[i]
```

```python
a = np.array([1 ,3])
xlim = [-4, 4]

scalars = np.random.uniform(low = xlim[0], high = xlim[1], size=100)


for i in range(len(scalars)):
    p = a * scalars[i]
    plt.plot(p[0],p[1],'ko')


plt.xlim(xlim)
plt.ylim(xlim)
plt.grid()
plt.show()
    
```

```python
pio.renderers.default = 'iframe'

a = np.array([3, 5, 1])
b = np.array([0, 2, 2])

scalars = np.random.uniform(low=-2, high=2, size=(100, 2))
points = np.zeros((100, 3))

for i in range(len(scalars)):
    points[i, :] = a * scalars[i, 0] + b * scalars[i, 1]

fig = go.Figure(data=[
    go.Scatter3d(
        x=points[:,0],
        y=points[:,1],
        z=points[:,2],
        mode='markers'
    )
])

fig.show() 
```
