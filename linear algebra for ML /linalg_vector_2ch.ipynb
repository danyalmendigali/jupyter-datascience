```python
def n(a):
    return np.sqrt(np.sum(a ** 2))

a = np.array([1, 2, 3])
norm = np.linalg.norm(a)
print(n(a))
print(norm)
```

### Ex 3 

```python
def ei(v):
    return (1 / np.linalg.norm(v) * v)

g = np.array([1, 2, 3])
g_norm1 = np.linalg.norm(ei(g))
print(ei(g))
print(g_norm1)
```

### Ex 8 

```python
a = np.array([3, 4])
b = np.array([3, 1])

beta = np.dot(a, b) / np.dot(a, a)
proj = beta * a
res = b - proj

print(res)

plt.figure(figsize=(6,6))
plt.axis('equal')
plt.grid(True)

plt.arrow(0, 0, a[0], a[1], width=0.2, head_width=0.5, length_includes_head=True)
plt.arrow(0, 0, b[0], b[1], width=0.2, head_width=0.5, length_includes_head=True)
plt.arrow(0, 0, proj[0], proj[1], width=0.2, head_width=0.5, length_includes_head=True)
plt.arrow(proj[0], proj[1], res[0], res[1], width=0.2, head_width=0.5, length_includes_head=True)


plt.xlim(-1, 6)
plt.ylim(-2, 6)

plt.show()
```

### Ex 9

```python
t = np.array([1, 5])
r = np.array([3, 5])

t_para = r * (np.dot(t, r) / np.dot(r, r))
t_perp = t - t_para
t_check = t_para + t_perp

print("t:", t)
print("t_para + t_perp:", t_check)

plt.figure(figsize=(6,6))
plt.axis('equal')
plt.grid(True)

# r
plt.arrow(0, 0, r[0], r[1], width=0.15, head_width=0.4, length_includes_head=True)

# t
plt.arrow(0, 0, t[0], t[1], width=0.15, head_width=0.4, length_includes_head=True)

# t_parallel
plt.arrow(0, 0, t_para[0], t_para[1], width=0.15, head_width=0.4, length_includes_head=True)

# t_perp (из конца t_para)
plt.arrow(t_para[0], t_para[1], t_perp[0], t_perp[1],
          width=0.15, head_width=0.4, length_includes_head=True)

plt.xlim(-2, 6)
plt.ylim(-2, 7)

plt.show()
```
