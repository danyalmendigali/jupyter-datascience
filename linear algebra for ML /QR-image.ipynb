```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('/Users/danialmendigali/jupyter-datascience/linear algebra for ML /DataLinal/img.jpg')
Q, R = np.linalg.qr(img)

Q_vis = (Q - Q.min()) / (Q.max() - Q.min())
R_vis = (R - R.min()) / (R.max() - R.min())

plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.title("Original Image")
plt.imshow(img, cmap='gray')
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Q Matrix")
plt.imshow(Q_vis, cmap='gray')
plt.axis("off")

plt.subplot(1,3,3)
plt.title("R Matrix")
plt.imshow(R_vis, cmap='gray')
plt.axis("off")

plt.show()
```
