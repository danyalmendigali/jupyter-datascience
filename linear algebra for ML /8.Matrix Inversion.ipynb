```python
import numpy as np
import tensorflow as tf
import torch
```

```python
x = np.array([[4, 2], [-5, -3]])
Xinv = np.linalg.inv(x)
Xinv
```

```python
y = np.array([4, -7])
y
```

```python
w = np.dot(Xinv, y)
w
```

```python
np.dot(x, w)
```

### In PyTorch and TensorFlow

```python
torch.inverse(torch.tensor([[4, 2], [-5, -3.]]))
```

```python
tf.linalg.inv(tf.Variable([[4, 2], [-5, -3.]]))
```
