```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt 
import torch
```

```python
a = np.array([[4, 2], [-5, -3]])
a
```

```python
lambdas, v = np.linalg.eig(a)
```

```python
v
```

```python
Vinv = np.linalg.inv(v)
Vinv
```

```python
Lambda = np.diag(lambdas)
Lambda
```

```python
np.dot(v, np.dot(Lambda, Vinv))
```
