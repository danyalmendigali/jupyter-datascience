```python
%%manim -qm CircleToSquare

class CircleToSquare(Scene):
    def construct(self):
        blue_circle = Circle(color=BLUE, fill_opacity=0.5)
        green_square = Square(color=GREEN, fill_opacity=0.8)
        self.play(Create(blue_circle))
        self.wait()
        
        self.play(Transform(blue_circle, green_square))
        self.wait()
```

```python
%%manim -qm VectorScalarMultiplication

from manim import *

class VectorScalarMultiplication(Scene):
    def construct(self):

        plane = NumberPlane()
        self.play(Create(plane))

        # исходный вектор
        v = Arrow(ORIGIN, [2,1,0], buff=0, color=YELLOW)
        label_v = Text("v").next_to(v.get_end(), RIGHT)

        self.play(Create(v), Write(label_v))
        self.wait()

        # текст
        scalar_text = Text("Multiply vector by 2").to_corner(UL)
        self.play(Write(scalar_text))
        self.wait()

        # новый вектор (в 2 раза длиннее)
        v2 = Arrow(ORIGIN, [4,2,0], buff=0, color=RED)
        label_v2 = Text("2v").next_to(v2.get_end(), RIGHT)

        self.play(
            Transform(v, v2),
            Transform(label_v, label_v2),
            run_time=2
        )

        self.wait()
```

```python
%%manim -qm MatrixVectorMultiplication

from manim import *

class MatrixVectorMultiplication(Scene):
    def construct(self):

        plane = NumberPlane()
        self.play(Create(plane))

        # исходный вектор
        v = Arrow(ORIGIN, [2,1,0], buff=0, color=YELLOW)
        label_v = Text("v").next_to(v.get_end(), RIGHT)

        self.play(Create(v), Write(label_v))
        self.wait()

        # текст матрицы (без LaTeX)
        matrix_text = Text("Matrix A = [[2,1],[1,1]]").to_corner(UL)
        self.play(Write(matrix_text))
        self.wait()

        # матрица
        matrix = [[2,1],
                  [1,1]]

        # новый вектор Av
        v2 = Arrow(ORIGIN, [5,3,0], buff=0, color=RED)
        label_v2 = Text("Av").next_to(v2.get_end(), RIGHT)

        # применение линейного преобразования
        self.play(
            ApplyMatrix(matrix, plane),
            Transform(v, v2),
            Transform(label_v, label_v2),
            run_time=3
        )

        self.wait()
```

```python
%%manim -qm EigenVectorsDemoNoLaTeX

from manim import *
import numpy as np

class EigenVectorsDemoNoLaTeX(Scene):
    def construct(self):
        # Настроим плоскость
        plane = NumberPlane(x_range=[-5,5], y_range=[-5,5])
        self.add(plane)
        self.wait(0.5)

        # Матрица для демонстрации
        A = np.array([[2, 1],
                      [1, 2]])

        origin = np.array([0,0,0])

        # Начальные векторы
        v1 = np.array([1,0])
        v2 = np.array([0,1])

        # Покажем начальные векторы
        arrow1 = Arrow(origin, v1, color=YELLOW, buff=0)
        arrow2 = Arrow(origin, v2, color=BLUE, buff=0)
        label1 = Text("v1").next_to(arrow1.get_end(), RIGHT)
        label2 = Text("v2").next_to(arrow2.get_end(), UP)
        self.play(Create(arrow1), Create(arrow2), Write(label1), Write(label2))
        self.wait(1)

        # Вычисляем собственные значения и векторы
        eigvals, eigvecs = np.linalg.eig(A)

        # Применим матрицу к собственным векторам
        transformed_v1 = A @ eigvecs[:,0]
        transformed_v2 = A @ eigvecs[:,1]

        # Покажем собственные векторы
        arrow_eig1 = Arrow(origin, eigvecs[:,0], color=RED, buff=0)
        arrow_eig2 = Arrow(origin, eigvecs[:,1], color=GREEN, buff=0)
        label_eig1 = Text("eigvec 1").next_to(arrow_eig1.get_end(), RIGHT)
        label_eig2 = Text("eigvec 2").next_to(arrow_eig2.get_end(), RIGHT)

        self.play(Transform(arrow1, arrow_eig1),
                  Transform(arrow2, arrow_eig2),
                  Transform(label1, label_eig1),
                  Transform(label2, label_eig2))
        self.wait(1)

        # Действие матрицы на собственные векторы
        arrow_eig1_transformed = Arrow(origin, transformed_v1, color=RED, buff=0)
        arrow_eig2_transformed = Arrow(origin, transformed_v2, color=GREEN, buff=0)

        self.play(
            Transform(arrow1, arrow_eig1_transformed),
            Transform(arrow2, arrow_eig2_transformed)
        )

        # Покажем собственные значения
        lambda1_label = Text(f"λ1 = {eigvals[0]:.2f}").next_to(arrow1.get_end(), RIGHT)
        lambda2_label = Text(f"λ2 = {eigvals[1]:.2f}").next_to(arrow2.get_end(), RIGHT)

        self.play(Write(lambda1_label), Write(lambda2_label))
        self.wait(2)
```

```python
%%manim -qm EigenVectors
from manim import *
import numpy as np

class EigenVectors(Scene):
    def construct(self):
        # Плоскость
        plane = NumberPlane(x_range=[-5,5], y_range=[-5,5])
        self.add(plane)
        self.wait(0.5)

        # Матрица
        A = np.array([[2,1],[1,2]])
        eigvals, eigvecs = np.linalg.eig(A)

        origin = np.array([0,0,0])

        # Преобразуем 2D -> 3D
        v1_3d = np.append(eigvecs[:,0], 0)
        v2_3d = np.append(eigvecs[:,1], 0)

        # Eigenvectors
        arrow1 = Arrow(origin, v1_3d, color=RED, buff=0)
        arrow2 = Arrow(origin, v2_3d, color=GREEN, buff=0)
        self.add(arrow1, arrow2)

        # Метки через MathTex
        label1 = MathTex("v_1").next_to(arrow1.get_end(), RIGHT)
        label2 = MathTex("v_2").next_to(arrow2.get_end(), RIGHT)
        lambda1_label = MathTex(f"\\lambda_1 = {eigvals[0]:.2f}").next_to(arrow1.get_end(), UP)
        lambda2_label = MathTex(f"\\lambda_2 = {eigvals[1]:.2f}").next_to(arrow2.get_end(), UP)

        self.play(Write(label1), Write(label2))
        self.play(Write(lambda1_label), Write(lambda2_label))
        self.wait(2)
```

```python
%%manim -qm SVD3Dto2DVectors
from manim import *
import numpy as np

class SVD3Dto2DVectors(Scene):
    def construct(self):
        plane = NumberPlane(x_range=[-5,5], y_range=[-5,5])
        self.add(plane)
        self.wait(0.5)

        # Исходные 3D-векторы
        v1 = np.array([2, 1, 3])
        v2 = np.array([1, 2, 1])
        vectors_3d = np.column_stack([v1, v2])  # shape (3,2)

        origin = np.array([0,0,0])

        # Сделаем стрелки в 3D, даже если проекция на XY
        arrow1 = Arrow(origin, np.array([v1[0], v1[1], 0]), color=BLUE, buff=0)
        arrow2 = Arrow(origin, np.array([v2[0], v2[1], 0]), color=GREEN, buff=0)
        label1 = Text("v1").next_to(arrow1.get_end(), RIGHT)
        label2 = Text("v2").next_to(arrow2.get_end(), RIGHT)
        self.play(Create(arrow1), Create(arrow2), Write(label1), Write(label2))
        self.wait(1)

        # Проекция на 2D через матрицу A (2x3)
        A = np.array([[1,0,0],
                      [0,1,0]])

        projected_vectors = A @ vectors_3d  # shape (2,2)

        # Сделаем 3D-стрелки из 2D векторов (добавляем z=0)
        arrow1_proj = Arrow(origin, np.array([projected_vectors[0,0], projected_vectors[1,0], 0]), color=RED, buff=0)
        arrow2_proj = Arrow(origin, np.array([projected_vectors[0,1], projected_vectors[1,1], 0]), color=ORANGE, buff=0)
        label1_proj = Text("v1'").next_to(arrow1_proj.get_end(), RIGHT)
        label2_proj = Text("v2'").next_to(arrow2_proj.get_end(), RIGHT)

        self.play(
            Transform(arrow1, arrow1_proj),
            Transform(arrow2, arrow2_proj),
            Transform(label1, label1_proj),
            Transform(label2, label2_proj)
        )
        self.wait(2)

        self.play(Write(Text("Проекция 3D → 2D через матрицу A").to_edge(UP)))
        self.wait(2)
```

```python
%%manim -qm SVD3DVisualization
from manim import *
import numpy as np

class SVD3DVisualization(ThreeDScene):
    def construct(self):
        # Настроим камеру для 3D
        self.set_camera_orientation(phi=75 * DEGREES, theta=-45 * DEGREES)

        # Оси
        axes = ThreeDAxes(x_range=[-5,5], y_range=[-5,5], z_range=[-5,5])
        self.add(axes)

        # Исходные 3D-векторы
        v1 = np.array([2, 1, 3])
        v2 = np.array([1, 2, 1])
        v3 = np.array([3, 0, 2])
        vectors = [v1, v2, v3]
        colors = [BLUE, GREEN, ORANGE]

        origin = np.array([0,0,0])

        # Создаём стрелки для исходных векторов
        arrows = []
        labels = []
        for i, vec in enumerate(vectors):
            arrow = Arrow3D(start=origin, end=vec, color=colors[i])
            label = Text(f"v{i+1}").scale(0.5).next_to(vec, RIGHT)
            arrows.append(arrow)
            labels.append(label)
            self.add(arrow, label)

        self.wait(1)

        # Создаем матрицу для SVD
        # Пример: A = U Σ V^T, пусть A 3x3
        A = np.array([[3,1,2],
                      [2,2,1],
                      [1,3,2]])

        # Выполним SVD
        U, S, VT = np.linalg.svd(A)

        # Показать разворот векторов в новое направление (V^T)
        rotated_vectors = VT @ np.column_stack(vectors)  # shape (3,3)
        arrows_rotated = []
        labels_rotated = []
        for i in range(3):
            end = rotated_vectors[:,i]
            arrow_rot = Arrow3D(start=origin, end=end, color=colors[i])
            label_rot = Text(f"v{i+1}'").scale(0.5).next_to(end, RIGHT)
            arrows_rotated.append(arrow_rot)
            labels_rotated.append(label_rot)

        self.play(
            *[Transform(arrows[i], arrows_rotated[i]) for i in range(3)],
            *[Transform(labels[i], labels_rotated[i]) for i in range(3)],
            run_time=2
        )
        self.wait(1)

        # Показать растяжение вдоль осей (S)
        stretched_vectors = U @ np.diag(S)  # shape (3,3)
        arrows_stretched = []
        labels_stretched = []
        for i in range(3):
            end = stretched_vectors[:,i]
            arrow_str = Arrow3D(start=origin, end=end, color=colors[i])
            label_str = Text(f"σ{i+1} v{i+1}").scale(0.5).next_to(end, RIGHT)
            arrows_stretched.append(arrow_str)
            labels_stretched.append(label_str)

        self.play(
            *[Transform(arrows_rotated[i], arrows_stretched[i]) for i in range(3)],
            *[Transform(labels_rotated[i], labels_stretched[i]) for i in range(3)],
            run_time=2
        )
        self.wait(1)

        # Проекция в 2D (удаляем z)
        arrows_2d = []
        labels_2d = []
        for i in range(3):
            end = np.array([stretched_vectors[0,i], stretched_vectors[1,i], 0])
            arrow_2d = Arrow3D(start=origin, end=end, color=colors[i])
            label_2d = Text(f"v{i+1}_2D").scale(0.5).next_to(end, RIGHT)
            arrows_2d.append(arrow_2d)
            labels_2d.append(label_2d)

        self.play(
            *[Transform(arrows_stretched[i], arrows_2d[i]) for i in range(3)],
            *[Transform(labels_stretched[i], labels_2d[i]) for i in range(3)],
            run_time=2
        )
        self.wait(2)
```

```python
from manim import *
import numpy as np

class SVDVisualization(Scene):
    def construct(self):
        # 1. Определяем матрицу A
        # Можно заменить на свою, но эта дает наглядный сдвиг и растяжение
        matrix_a = np.array([[2, 1], [1, 1]])
        
        # Вычисляем SVD через numpy
        U, S, Vh = np.linalg.svd(matrix_a)
        V = Vh.T
        
        # Настройка сетки
        grid = NumberPlane()
        circle = Circle(radius=1, color=WHITE)
        basis_i = Vector([1, 0], color=GREEN)
        basis_j = Vector([0, 1], color=RED)
        
        labels = VGroup(
            Text("SVD: A = U Σ Vᵀ", font_size=36).to_edge(UP),
            Tex(r"$\vec{x}$", color=WHITE).next_to(circle, DR)
        )

        self.add(grid, circle, basis_i, basis_j, labels)
        self.play(Write(labels))
        self.wait()

        # --- ШАГ 1: Применяем V^T (Первый поворот) ---
        # В SVD первый шаг — это поворот векторами V
        self.play(
            circle.animate.apply_matrix(Vh),
            basis_i.animate.apply_matrix(Vh),
            basis_j.animate.apply_matrix(Vh),
            run_time=2
        )
        self.play(Indicate(labels[0]))
        self.wait()

        # --- ШАГ 2: Применяем Sigma (Растяжение по осям) ---
        # Sigma — это диагональная матрица сингулярных чисел
        sigma_matrix = np.diag(S)
        self.play(
            circle.animate.apply_matrix(sigma_matrix),
            basis_i.animate.apply_matrix(sigma_matrix),
            basis_j.animate.apply_matrix(sigma_matrix),
            run_time=2
        )
        self.wait()

        # --- ШАГ 3: Применяем U (Второй поворот) ---
        self.play(
            circle.animate.apply_matrix(U),
            basis_i.animate.apply_matrix(U),
            basis_j.animate.apply_matrix(U),
            run_time=2
        )
        
        final_text = Text("Трансформация завершена", font_size=32, color=YELLOW).to_edge(DOWN)
        self.play(Write(final_text))
        self.wait(2)
```
