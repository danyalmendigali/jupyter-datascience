```python
df.loc[df.sample(frac=0.1).index, 'age'] = np.nan
df.loc[df.sample(frac=0.1).index, 'bmi'] = np.nan
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.shape[0]
```

```python
df.dropna(inplace=True)
```

```python
df.fillna({
    'age' : df['age'].mean(),
    'bmi' : df['bmi'].max()
}, inplace=True)
```

```python
df.loc[
    (df['age'] == df['age'].mean())
].head(5)
```

```python
df.loc[df.sample(frac=0.08).index, 'sex'] = np.nan
df.loc[df.sample(frac=0.08).index, 'smoker'] = np.nan
df.loc[df.sample(frac=0.08).index, 'region'] = np.nan
```

```python
df.isnull().sum()
```

```python
df['region'].mode()[0]
```

```python
df.loc[
    (df['region'] == 'southeast')
].shape[0]
```

```python
df.fillna({
    'sex' : df['sex'].mode()[0],
    'smoker' : 'unknown',
    'region' : df['region'].mode()[0]
}, inplace=True)
```

```python
df.head(5)
```

```python
df['risk_level'] = df.apply(
    lambda x: 'high' if (x['smoker'] == 'yes' and x['bmi'] > 30) or (x['smoker'] == 'yes' and x['age'] >= 55)
    else('medium' if x['smoker'] == 'yes' and 25 <= x['bmi'] <= 30 else 'low'),
    axis=1
)
```

```python
df.loc[
    (df['smoker'] == 'yes')
].head(5)
```

```python
def family_status(r):
    if r['children'] >= 3:
        return 'large_family'
    elif r['children'] >= 1 and r['children'] <= 2:
        return 'small_family'
    else:
        return 'no_family'
        
df['family_status'] = df.apply(family_status, axis=1)
```

```python
df[['children', 'family_status']].head(5)
```

```python
df['bmi_category'] = np.where(
    df['bmi'] < 18.5, 'underweight',
    np.where(df['bmi'] < 25, 'normal', 'overweight')
)
```

```python
df.sort_values(by='bmi', ascending=False).head(5)
```

```python
df['high_charges'] = np.where(
    df['charges'] > df['charges'].mean(), 1, 0
)
```
