```python
import numpy as np 
import pandas as pd 
```

### Concat / merge 

```python
students = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "name": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "age": [18, 19, 18, 20, 17]
})


scores = pd.DataFrame({
    "id": [1, 2, 2, 4, 6], 
    "math": [90, 75, 88, 60, 95],
    "english": [85, 80, 78, 70, 88]
})


schools = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "city": ["Almaty", "Astana", "Almaty", "Shymkent", "Astana"],
    "school": ["NIS", "NIS", "RFMSH", "NIS", "Lyceum"]
})
```

```python
pd.concat([students, scores, schools])
```

```python
pd.merge(students, scores, on="id", how="inner") # Берёт только те строки, где id есть в ОБЕИХ таблицах.
```

```python
pd.merge(students, scores, on="id", how="left")
```

```python
pd.merge(students, scores, on="id", how="right")
```

```python
d1 = pd.merge(students, scores, on='id', how='outer')
d1
```

```python
pd.merge(d1, schools, on='id', how='outer')>
```

```python
pd.concat([schools, scores], axis=1)
```

```python
pd.concat([schools, scores], axis=0)
```

```python
pd.merge(students, scores, on='id', how='inner')
```

```python
pd.merge(students, scores, on='id', how='right')
```

```python
pd.merge(students, scores, on='id', how='outer')
```

```python
df = pd.concat([students, scores, schools], axis=1)
```
