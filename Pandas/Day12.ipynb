### Optimization and vectorization

```python
df = pd.read_csv('/Users/danialmendigali/jupyter-datascience/Pandas/data set /insurance.csv')
```

```python
df['sex'].value_counts()
```

```python
def risk_level(row):
    if row['smoker'] == 'yes' and row['bmi'] > 30:
        return 'high'
    elif row['smoker'] == 'yes':
        return 'medium'
    else:
        return 'low'

df['risk'] = df.apply(risk_level, axis=1)
```

```python
df.head()
```

```python
df['risk'] = np.where(
    (df['smoker'] == 'yes') & (df['bmi'] > 30), 'high',
    np.where(df['smoker'] == 'yes', 'medium', 'low')
)
```

### apply = convenient when the logic is complex, but slow.

```python
over_50 = []
for age in df['age']:
    if age > 50:
        over_50.append(1)
    else:
        over_50.append(0)

df['over_50'] = over_50
```
