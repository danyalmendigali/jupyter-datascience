```python
import pandas as pd 
import numpy as np
import math
```

```python
a = np.array([1, 2, 3])
b = np.array([1, -2, -3])
```

```python
## vector length
l = math.sqrt(a[0] ** 2 + a[1] ** 2 + a[2] ** 2)
l1 = np.linalg.norm(a)
print(l)
print(l1)
```

```python
## cosine similarity 
f = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
print(f)
```

```python
def depend(matrix):
    rank = np.linalg.matrix_rank(matrix) 
    if rank < len(matrix): # dependent
        return True
    else:
        return False  # undependent

v1 = np.array([1, 2, 3])
v2 = np.array([2, 4, 6])  # v2 = 2*v1
v3 = np.array([0, 1, 1])

matrix = np.array([v1, v2, v3])

if depend(matrix):
    print("True")
else:
    print("False")
```

```python
v1 = np.array([1, 0, 0])
v2 = np.array([0, 1, 0])
v3 = np.array([0, 0, 1])

matrix = np.array([v1, v2, v3])

if depend(matrix):
    print("True")
else:
    print("False")
```
