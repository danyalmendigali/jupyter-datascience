```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd 
```

### Vectors in linear algebra 

```python
a = np.array([1.72, 54])
b = np.array([1.56, 100])

plt.figure()

plt.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1)
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1)

plt.xlim(0, 3)
plt.ylim(0, 110)

cos_similarity = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle_rad = np.arccos(cos_similarity)
angle_degrees = np.degrees(angle_rad)

print(cos_similarity, angle_rad, angle_degrees)

plt.show()
```

```python
a = np.array([1.72, 54, 36.2])
b = np.array([1.56, 47, 30.0])

cos_similarity = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
angle_rad = np.arccos(cos_similarity)
angle_degrees = np.degrees(angle_rad)

print(cos_similarity, angle_rad, angle_degrees)
```

```python
# vector addition
a = np.array([3, 2])
b = np.array([2, -1])
c = a + b  ## ([5, 1])
plt.figure()
plt.quiver(0, 0, c[0], c[1], angles='xy', scale_units='xy', scale=1)
plt.xlim(0, 7)
plt.ylim(0, 7)
plt.grid(True)
plt.show()

print(c)
```

```python
# vector subtraction
a = np.array([3, 2])
b = np.array([2, -1])
c = a - b  ## ([5, 1])
plt.figure()
plt.quiver(0, 0, c[0], c[1], angles='xy', scale_units='xy', scale=1)
plt.xlim(0, 7)
plt.ylim(0, 7)
plt.grid(True)
plt.show()

print(c)
```

```python
# multiplication of a vector by a number
a = np.array([3, 2])
b = 2.0 * a 
plt.figure()
plt.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1)
plt.xlim(0, 7)
plt.ylim(0, 7)
plt.grid(True)
plt.show()

print(b)
```

## Euclidean distance

```python
v1 = np.array([5, 6, 7, 8, 9])
v_norm = np.linalg.norm(v1)  ## norm/length vector
print(v_norm)
```

```python
v1 = np.array([[1, 0], [1, 1]])
np.linalg.eig(v1)

unity_vector = np.array([1, 0, 0])
unity_vector = unity_vector / v_norm
unity_vector
```
