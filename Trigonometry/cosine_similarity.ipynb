```python
### Creates an array of 100 numbers, evenly distributed: from 0 to 2π.
x = np.linspace(0, 2*np.pi, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)
plt.plot(x, y_sin, color='green')
plt.plot(x, y_cos, color='blue')
```

```python
# only dots
plt.scatter(x, y_cos)
plt.scatter(x, y_sin)
plt.show()
```

```python
# Dot product
A = np.array([1, 2])
B = np.array([2, 5])
dot = np.dot(A, B)
print(dot)
```

```python
cos_sim = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))
print(cos_sim)
```

```python
A = np.array([1, 2, 3])
B = np.array([2, 4, 6])

# dot product
dot_product = np.dot(A, B)
print("Dot product:", dot_product)  # 1*2 + 2*4 + 3*6 = 28

# vector lengths
norm_A = np.linalg.norm(A)  # sqrt(1^2+2^2+3^2) = sqrt(14)
norm_B = np.linalg.norm(B)  # sqrt(2^2+4^2+6^2) = sqrt(56)

# cosine similarity
cos_sim = dot_product / (norm_A * norm_B)
print("Cosine similarity:", cos_sim)
```
